#### a) escribe el pseudocodigo del programa

~~~
Inicio
    Leer x
    b <- x
    Mientras b != (x / b) Hacer
        b <- 0.5 * ((x / b) + b)
    Fin Mientras
    Escribir b
Fin

##### Sección 1.1: Implementación del Algoritmo Babilónico
- Implementación fiel al diagrama de flujo
    Este código sigue estrictamente la lógica presentada en el planteamiento inicial:

```python
x = float(input("Ingresa x: "))
b = x

# Condición de parada según el diagrama de flujo original
while b != (x / b):
    b = 0.5 * ((x / b) + b)

print(f"Resultado exacto: {b}")
```

##### B. Optimización para entornos de producción
- En la práctica, debido a la precisión de los números de punto flotante (floats), la igualdad exacta != puede generar bucles infinitos en números sin raíz exacta. Por ello, se recomienda el uso de un margen de error o epsilon:

```python
def raiz_cuadrada_optimizada(x, epsilon=1e-10):
    b = x
    # Verificación de diferencia mínima para asegurar la convergencia
    while abs(b - (x / b)) > epsilon:
        b = 0.5 * ((x / b) + b)
    return b

print(f"Resultado optimizado: {raiz_cuadrada_optimizada(x)}")
```

#### b) que funcion realiza el programa

El programa realiza el cálculo de la raíz cuadrada ($\sqrt{x}$) de un número dado $x$. 
Utiliza un algoritmo numérico iterativo conocido como el Método Babilónico

La lógica se basa en los siguientes puntos:
- **punto de equilibrio** El bucle Mientras se detiene cuando se cumple la condición $b = x / b$. Matemáticamente, si despejamos la variable $b$, esto equivale a $b^2 = x$, lo que significa que $b$ es la raíz exacta de $x$
- **Aproximación sucesiva** En cada iteración, el programa calcula un nuevo valor para $b$ promediando el valor actual con $x/b$ mediante la fórmula $b \leftarrow \frac{1}{2}(\frac{x}{b} + b)$. Esto hace que $b$ se acerque muy rápidamente al valor real de la raíz cuadrada.

#### c) Complejidad del algoritmo ($O(g)$)
La complejidad de este algoritmo es $O(\log(\log(1/\epsilon)))$, donde $\epsilon$ es el nivel de precisión deseado.

En términos prácticos, se dice que tiene convergencia cuadrática. Esto significa que el número de dígitos correctos de la solución se duplica aproximadamente en cada iteración, lo que lo hace mucho más eficiente que otros métodos de búsqueda.

La notación Big O mide el crecimiento del esfuerzo computacional. En el caso del método babilónico, es altamente eficiente porque su convergencia es cuadrática, lo que significa que alcanza una precisión de nivel científico en un número mínimo de iteraciones, independientemente de qué tan grande sea el número de entrada.

### 2. Escribe un pseudocodigo recursivo y otro iterativo para calcular el factorial de un numero dado.

#### iterativo
~~~
inicio
    leer x
    factorial <- 1
    para index <- 1 hasta x Hacer
        factorial <- factorial * index
    fin para
    escribir factorial
fin
~~~

```python
def calcular_factorial(n):
    factorial = 1
    for index in range(1, n + 1):
        factorial = factorial * index
    return factorial

# Prueba del algoritmo
x = int(input("De qué número quieres el factorial: "))
print(f"El factorial de {x} es: {calcular_factorial(x)}")
```

#### recursivo
~~~
funcion factorial(n)
    si n == 0 o n==1 Entonces 
        devolver 1
    sino
        devolver n * factorial(n-1)
    fin si
fin funcion
inicio
    leer x
    escribir factorial(x)
fin
~~~

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

x=input("numero a factorializar: ")
print(factorial(int(x)))
```

#### a) Describa las diferencias. 
1. **Mecanismo de Control**
   - **Iterativo**: Utiliza estructuras de control de repetición (bucles) como Para (for) o Mientras (while). El proceso se repite en un orden lineal hasta que la condición del bucle deja de cumplirse.
   - **Recursivo**: Utiliza llamadas a la propia función. El proceso se divide en "capas" o instancias de la misma función que se abren una dentro de otra.
2. **Gestión de la Memoria** (La diferencia más importante)
    - **Iterativo**: Es muy eficiente en memoria. Solo utiliza unas cuantas variables locales (como factorial e index) que se van actualizando en el mismo espacio de memoria. Su complejidad de espacio es $O(1)$.
    - **Recursivo**: Cada vez que la función se llama a sí misma, la computadora crea un "marco de activación" en una estructura llamada Pila de Ejecución (Stack). Esto guarda el estado de la función, sus variables y la dirección de retorno. Si calculas el factorial de 1,000, tendrás 1,000 marcos abiertos en memoria al mismo tiempo. Su complejidad de espacio es $O(n)$.
3. **Condición de Parada**
    - **Iterativo**: El ciclo termina cuando el contador alcanza el límite establecido (el valor de $x$).
    - **Recursivo**: Depende estrictamente de un Caso Base (cuando $n=1$ o $n=0$). Si este caso no se define correctamente, el programa entra en una recursión infinita y colapsa.

#### b) ¿Cuál de ellos es mas eficiente y por que? 

El método iterativo es más eficiente.

¿Por qué?

- **Menos consumo de memoria**: Al no utilizar la pila de llamadas (stack), no hay riesgo de sufrir un error de Stack Overflow (desbordamiento de pila) si el número de entrada es muy grande.

- **Velocidad de procesamiento**: Cada llamada recursiva implica un "gasto extra" (overhead) para la CPU, ya que debe crear nuevos marcos de función, pasar argumentos y gestionar direcciones de retorno. Un bucle simple es una operación mucho más directa y rápida para el procesador.

### 1. Introducción: Fases del otorgamiento de crédito
En una institución financiera, el proceso de otorgamiento de crédito es un ciclo estructurado para mitigar el riesgo de impago. Se divide generalmente en estas fases:

- **Prospección y Solicitud**: El cliente inicia el proceso proporcionando sus datos personales, financieros y el propósito del crédito.

- **Evaluación de Riesgo (Credit Scoring)**: Es la fase más crítica. Aquí se utilizan modelos matemáticos y de Machine Learning para analizar el historial del cliente (Buró de Crédito) y su capacidad de pago. Se asigna una "calificación" o score.

- **Verificación y Validación**: Se comprueba la veracidad de la información y documentos entregados (identidad, comprobantes de ingresos, etc.).

- **Decisión y Aprobación**: Con base en el score y las políticas de riesgo, el sistema (o un comité) decide si aprueba, rechaza o condiciona el monto y la tasa de interés.

- **Instrumentación y Desembolso**: Se formaliza el contrato legalmente y se transfieren los fondos al cliente.

- **Seguimiento y Cobranza**: Se monitorea el comportamiento de pago para detectar riesgos tempranos de mora.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import GridSearchCV
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
statlog_german_credit_data = fetch_ucirepo(id=144) 
  
# data (as pandas dataframes) 
X = statlog_german_credit_data.data.features 
y = statlog_german_credit_data.data.targets 
  
# metadata 
print(statlog_german_credit_data.metadata) 
  
# variable information 
print(statlog_german_credit_data.variables) 
```

```python
# Diccionario completo de mapeo
diccionario_nombres = {
    'Attribute1': 'status_cuenta',
    'Attribute2': 'duracion_meses',
    'Attribute3': 'historial_crediticio',
    'Attribute4': 'proposito',
    'Attribute5': 'monto_credito',
    'Attribute6': 'ahorros',
    'Attribute7': 'tiempo_empleo',
    'Attribute8': 'tasa_pago_ingresos',
    'Attribute9': 'estado_civil_sexo',
    'Attribute10': 'otros_deudores',
    'Attribute11': 'tiempo_residencia',
    'Attribute12': 'propiedad',
    'Attribute13': 'edad',
    'Attribute14': 'otros_planes_pago',
    'Attribute15': 'vivienda',
    'Attribute16': 'creditos_existentes',
    'Attribute17': 'trabajo',
    'Attribute18': 'dependientes',
    'Attribute19': 'telefono',
    'Attribute20': 'trabajador_extranjero'
}

# Aplicar el cambio a X
X.rename(columns=diccionario_nombres, inplace=True)

# También es buena idea renombrar la variable objetivo 'y'
y.rename(columns={'target': 'riesgo_crediticio'}, inplace=True)

print("¡Nombres actualizados!")
print(X.columns)
```

```python
# Ver los primeros 5 registros
print(X.head())
```

Para que los algoritmos puedan procesar la información, se aplicó codificación de variables categóricas (One-Hot Encoding / Variables Dummy) y se escalaron las variables numéricas con StandardScaler para evitar que variables como el Monto del Crédito dominaran por su escala.

```python
# 1. Transformar categorías a números (0 y 1)
# Esto "explota" el DataFrame: de 20 columnas pasaremos a unas 60 aproximadamente
X_cleaned = pd.get_dummies(X)

# 2. Escalamiento
# Usamos StandardScaler para que la media sea 0 y la desviación estándar 1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cleaned)

# Convertimos X_scaled de nuevo a un DataFrame para que sea fácil de manejar
X_scaled_df = pd.DataFrame(X_scaled, columns=X_cleaned.columns)

print("¡Limpieza y preparación completada!")
print(f"Nuevas dimensiones: {X_scaled_df.shape}")
X_scaled_df.head()
```

```python
print(X_cleaned.head())
```

se obcerba que las columnas nuevas tienen valores True o False.

Dato técnico: La mayoría de los modelos de scikit-learn (como PCA o Random Forest) pueden interpretar esto como 1 y 0 automáticamente pero para evitar cualquier advertencia (warning) y que los datos se vean más limpios, se convertiran de una vez a numeros enteros de forma que true y false cambiaran a 1 y 0 respectivamente.

```python
# 1. Escalamiento (Obligatorio para PCA)
# Esto hace que todas las columnas tengan media 0 y varianza 1
"""
scaler = StandardScaler()
# X_scaled = scaler.fit_transform(X_cleaned)
"""
# 2. Aplicar PCA
# Vamos a reducir las 61 columnas a solo 2 para poder graficarlas
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Convertimos a DataFrame para que sea fácil de manejar
df_pca = pd.DataFrame(data=X_pca, columns=['Componente 1', 'Componente 2'])

# 3. Graficar el resultado
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Componente 1', y='Componente 2', data=df_pca, alpha=0.7)
plt.title('Punto 3.1: Reducción de Dimensionalidad (PCA)')
plt.xlabel(f'Componente 1 ({pca.explained_variance_ratio_[0]*100:.1f}% de varianza)')
plt.ylabel(f'Componente 2 ({pca.explained_variance_ratio_[1]*100:.1f}% de varianza)')
plt.grid(True)
plt.show()

print(f"Varianza total explicada por los 2 componentes: {pca.explained_variance_ratio_.sum() * 100:.2f}%")
```

#### 3.1. 

Se realizó la reducción de dimensionalidad mediante PCA. Los dos primeros componentes principales capturan el 11.5% de la varianza total. Aunque este valor es bajo, es consistente con datasets financieros de alta dimensionalidad (61 variables tras el encoding), donde la información está distribuida entre múltiples factores socio-demográficos y financieros. Esta proyección permite visualizar la complejidad del problema y sirve como base para el algoritmo de clustering.


```python
inercia = []
for k in range(1, 11):
    kmeans_test = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_test.fit(X_scaled)
    inercia.append(kmeans_test.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inercia, marker='o')
plt.title('Método del Codo')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Inercia')
plt.grid(True)
plt.show()
```

#### 3.2 **Interpretación del Método del Codo (WCSS)**
Para determinar el número óptimo de segmentos de clientes, aplicamos el Método del Codo (Elbow Method).
Este método calcula la Inercia o WCSS (Within-Cluster Sum of Squares), que mide la suma de las distancias al cuadrado entre cada punto y el centro (centroide) de su cluster.

##### **Análisis de la gráfica**:

**Punto de Quiebre**: Observamos que la inercia disminuye drásticamente a medida que aumentamos el número de clusters de 1 a 3.

**Rendimientos Decrecientes**: A partir de $k=3$, la curva comienza a aplanarse notablemente. Esto indica que añadir un cuarto o quinto grupo no aporta una mejora significativa en la cohesión de los datos, pero sí aumentaría la complejidad del modelo.

**Decisión Técnica**:Se selecciona $K = 3$ como el número óptimo de clusters. Esta elección nos permite obtener grupos lo suficientemente compactos y distintos entre sí, manteniendo un nivel de granularidad que es interpretable para el negocio. Con esta configuración, el algoritmo es capaz de capturar las tres realidades principales de la cartera de clientes: el segmento joven, el segmento de estabilidad intermedia y el segmento premium.

```python
from sklearn.cluster import KMeans

# Buscamos 3 grupos
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Agregamos los grupos al gráfico para ver la separación
df_pca['cluster'] = clusters

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Componente 1', y='Componente 2', hue='cluster', data=df_pca, palette='Set1')
plt.title('Segmentación de Clientes (Clusters)')
plt.show()
```

3.4 Realizar un comparativo con dos modelos de aprendizaje supervisado 

preparamos la variable objetivo

```python
# Convertimos: 1 (Bueno) -> 1, 2 (Malo) -> 0
# Así el modelo predice la "probabilidad de ser buen pagador"
y_binaria = y.iloc[:, 0].map({1: 1, 2: 0})
```

dividimos los datos para entrenar y para testear
no se puede evaluar un modelo con los mismos datos que usaste para entrenarlos entonces separaremos un 20% para el test

```python
# Dividimos los datos ya escalados (X_scaled) que obtuvimos antes
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_binaria, test_size=0.2, random_state=42, stratify=y_binaria
)

print(f"Entrenamiento: {X_train.shape[0]} registros")
print(f"Prueba (Examen): {X_test.shape[0]} registros")
```

elegimos bosque aleatorios o random forest es uno de los mas potentes para este tipo de casos 

```python
# Creamos y entrenamos el modelo
modelo_rf = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_rf.fit(X_train, y_train)

# Hacemos las predicciones sobre el set de prueba
y_pred = modelo_rf.predict(X_test)
```

evaluamos con una matriz de confusion 

```python
# Sacamos la matriz
cm = confusion_matrix(y_test, y_pred)

# La graficamos
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Malo (0)', 'Bueno (1)'], 
            yticklabels=['Malo (0)', 'Bueno (1)'])
plt.xlabel('Predicción del Modelo')
plt.ylabel('Realidad (Cliente)')
plt.title('Matriz de Confusión - Random Forest')
plt.show()

# Reporte de precisión
print(classification_report(y_test, y_pred))
```

1. Interpretación de los Resultados
Accuracy (0.74): El modelo acierta 3 de cada 4 casos. Es un buen desempeño inicial para un problema tan complejo.

   - **Clase 1 (Buenos Pagadores)**: Tienes un Recall de 0.88. Esto es excelente; significa que el modelo es muy bueno encontrando a los clientes que sí pagan. no perdería muchos clientes buenos.

   - **Clase 0 (Malos Pagadores)**: Aquí está el reto. El Recall es de 0.43. Significa que de cada 10 personas que NO van a pagar, el modelo solo detecta a 4. Las otras 6 están pasando como "buenos" (Falsos Positivos).

1. El Impacto en el Negocio
En el otorgamiento de crédito, un Falso Positivo (decirle que sí a alguien que no va a pagar) es mucho más caro que un Falso Negativo (decirle que no a alguien que sí pagaría).

**El modelo actual es "optimista"**: Prefiere aprobar créditos. Para mejorar esto en una segunda fase, sugeriria "ajustar el umbral de decisión" para ser más estrictos con los de Clase 0.

Punto 3.4: Comparación de Modelos
se realizara un regresion logistica 

```python
# 1. Entrenar Regresión Logística
modelo_lr = LogisticRegression(random_state=42)
modelo_lr.fit(X_train, y_train)

# 2. Predicciones
y_pred_lr = modelo_lr.predict(X_test)

# 3. Comparar Accuracy
acc_rf = accuracy_score(y_test, y_pred)
acc_lr = accuracy_score(y_test, y_pred_lr)

print(f"Accuracy Random Forest: {acc_rf:.2f}")
print(f"Accuracy Regresión Logística: {acc_lr:.2f}")

# Ver si la Regresión Logística es mejor detectando a los "Malos" (Clase 0)
print("\nReporte Regresión Logística:")
print(classification_report(y_test, y_pred_lr))
```

- **Eficiencia de Ventas**: El Random Forest es excelente para el área comercial, ya que identifica al 88% de los buenos clientes, minimizando el rechazo injustificado de créditos.

**Recomendación Final**:
Se recomienda el uso del Random Forest debido a su mayor robustez y Accuracy. Sin embargo, dado que el costo de un 'Falso Positivo' (aprobar a un mal pagador) es alto, se sugiere implementar un ajuste de umbral (threshold) en el modelo para mejorar la detección de morosos sin perder la alta precisión que ofrece este algoritmo."

```python
# 1. Definimos los parámetros que queremos probar
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5, 10],
    'criterion': ['gini', 'entropy']
}

# 2. Creamos el buscador (GridSearch)
# cv=5 significa que dividirá los datos en 5 partes para validar internamente
grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42), 
                           param_grid=param_grid, 
                           cv=5, n_jobs=-1, verbose=1, scoring='accuracy')

# 3. Entrenamos buscando la mejor combinación
grid_search.fit(X_train, y_train)

print(f"Mejores parámetros encontrados: {grid_search.best_params_}")

# 4. Usamos el mejor modelo para predecir
mejor_modelo_rf = grid_search.best_estimator_
y_pred_mejorado = mejor_modelo_rf.predict(X_test)
```

```python
# 1. Obtener las probabilidades de "ser buen pagador" (Clase 1)
probabilidades = mejor_modelo_rf.predict_proba(X_test)[:, 1]

# 2. Aplicar un umbral más estricto (0.7 en lugar de 0.5)
umbral_estricto = 0.7
y_pred_estricto = (probabilidades > umbral_estricto).astype(int)

# 3. Ver el impacto en la detección de morosos
print(f"--- Evaluación con Umbral de Negocio ({umbral_estricto}) ---")
print(classification_report(y_test, y_pred_estricto))

# 4. Graficar la Matriz de Confusión Final
plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, y_pred_estricto), annot=True, fmt='d', cmap='Greens')
plt.title('Matriz Final: Enfoque en Mitigación de Riesgo')
plt.xlabel('Predicción (0:Malo, 1:Bueno)')
plt.ylabel('Realidad')
plt.show()
```

Tras el ajuste del umbral a 0.7, el modelo prioriza la mitigación de riesgo. Se logró incrementar la detección de clientes con perfil de impago (Clase 0) en un 37%, garantizando que la cartera aprobada tenga una **precisión del 88%**. Este balance protege la liquidez de la empresa frente a la incertidumbre del mercado.

```python
# 1. Extraer la importancia de las variables del modelo final
importancias = mejor_modelo_rf.feature_importances_
nombres_variables = X_cleaned.columns  

# 2. Crear un DataFrame para facilitar la manipulación y filtrado
df_importancia = pd.DataFrame({
    'Variable': nombres_variables,
    'Importancia': importancias
})

# 3. Ordenar de mayor a menor y tomar las 10 más críticas
df_importancia = df_importancia.sort_values(by='Importancia', ascending=False).head(10)

# 4. Crear la visualización
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Importancia', 
    y='Variable', 
    data=df_importancia, 
    palette='viridis' 
)

# 5. Estética y títulos
plt.title('Top 10 Variables Críticas para la Predicción de Riesgo', fontsize=14, fontweight='bold')
plt.xlabel('Nivel de Importancia (Contribución al Modelo)', fontsize=12)
plt.ylabel('Atributo del Cliente', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

El modelo nos revela que las variables más críticas para predecir un impago son monto credito y duracion meses. Esto indica a la institución que las políticas de riesgo deben ser mucho más estrictas en préstamos a largo plazo o de montos elevados.

Nivel de Endeudamiento Externo (Buró): Integrar el número de líneas de crédito activas en otras instituciones y el porcentaje de utilización de las mismas.

Comportamiento Digital (App/Web): Variables como el tiempo de llenado de la solicitud, dispositivo utilizado o frecuencia de consulta del buró, las cuales en modelos alternativos de scoring han demostrado estar fuertemente correlacionadas con la prevención de fraudes y el nivel de riesgo.
